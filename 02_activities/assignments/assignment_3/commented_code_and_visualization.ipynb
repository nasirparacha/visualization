```python
 import matplotlib.pyplot as plt
 import seaborn as sns
 import pandas as pd
```

**link to dataset** 

https://open.toronto.ca/dataset/apartment-building-registration/

```python
# import csv dataset

df = pd.read_csv('/content/Apartment Building Registration Data.csv')
```

```python
# getting informationabout first five rows
df.head(5)
```

```python
# checking non-null values, data types
df.info()
```

```python
# drop null values APPROVED_FIRE_SAFETY_PLAN

df['APPROVED_FIRE_SAFETY_PLAN'].dropna(inplace=True)
df.info() # checking final dataset
```

```python
# create pie chartfor APPROVED_FIRE_SAFETY_PLAN

approval_counts = df['APPROVED_FIRE_SAFETY_PLAN'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(approval_counts, labels=approval_counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('viridis', len(approval_counts)))
plt.title('Distribution of APPROVED_FIRE_SAFETY_PLAN')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```
